# Basic usage

Following example shows how Falcondale SDK could be used to benchmark some known QML approaches just by selecting a Pandas Dataframe. We will start by getting a sample dataset from scikit-learn.

```python
from sklearn import datasets

# import some data to play with
X, y = datasets.load_breast_cancer(return_X_y=True, as_frame=True)
X
```

We can check it is a binary classification task.

```python
y.unique()
```

Being all numerical datatypes it is a simple case to be explored with Falcondale SDK. Falcondale SDK should be able to ease that task going into different functionalities as we will explore this binary classification use case.

```python
from falcondale import Project

dataset = X.copy()
dataset["target"] = y

myproject = Project(dataset, target="target")
```

```python
myproject.list_options()
```

We need to preprocess the information so that missing values are filled and the dataset is scaled, which is important when encoding data into quantum states. One important feature is also the ability to reduce dimensionality. Let's start just by simply going for the default options.

```python
myproject.preprocess()
myproject.show_features()
```

One of the first things we could do is try to reduce the size of our dataset. We will try to perform a Quantum Feature Selection and see what it says.

```python
features = myproject.feature_selection(max_cols=3)
myproject.show_features()
```

This also modifies the internal dataset so that those features are going to be considered. Some simple classification tasks can be invoked thanks to the already identified target column.

```python
model = myproject.evaluate("qsvc")
model.print_report()
```

```python
model.list_metrics()
```

```python
model.metric("auc")
```

Is it any good? A fair question when working with quantum models is how they compare against a classical counterpart. The `svc` option trains a classical Support Vector Classifier on the very same features.

```python
baseline = myproject.evaluate("svc")
print(f"QSVC AUC: {model.metric('auc'):.3f} | SVC AUC: {baseline.metric('auc'):.3f}")
```

Models can be saved and retrieved afterwards. Make sure you use the same dataset structure when using the prediction function.

```python
myproject.save_model(model, "my_qsvc.model")
```

```python
model = myproject.load_model("my_qsvc.model")
```

```python
from sklearn.metrics import classification_report

y_pred = model.predict(X)

print(classification_report(y, y_pred))
```
