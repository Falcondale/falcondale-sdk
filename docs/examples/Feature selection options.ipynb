# Quantum Feature Selection

Feature selection posed as an Ising type of problem can be tackled by many different options. As it was mentioned in the documentation, the idea is to define each feature importance as its importance and the quadratic relation weight as the redundancy between different pairs available.

![featureselection.png](../../assets/featureselection.png)

This what Falcondale builds and then offers different options to solve it.

```python
from sklearn import datasets

# import some data to play with
X, y = datasets.load_breast_cancer(return_X_y=True, as_frame=True)
X
```

```python
from falcondale import Project

dataset = X.copy()
dataset["target"] = y

myproject = Project(dataset, target="target")
myproject.preprocess()
```

By default, **Simulated Annealing** will be the technique chose to select our feature subset.

```python
features = myproject.feature_selection(max_cols=10)
features
```

```python
myproject.show_features()
```

We can try to train a model and check how it would end up being with this selection.

```python
model = myproject.evaluate("qnn")
model.print_report()
```

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=10, method="sb")
features
```

Both methods tackle the very same QUBO, so they tend to agree on easy cases. Let's make it harder by reducing the number of features further.

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=3, method="sb")
features
```

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=3, method="sa")
features
```

Both solvers are forced to pick exactly the requested number of features thanks to the penalty term in the QUBO, but the selected subsets may differ. Let's check their performance when training a model.

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=3, method="sb")
model = myproject.evaluate("qsvc")
model.print_report()
```

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=3, method="sa")
model = myproject.evaluate("qsvc")
model.print_report()
```

Finally, **Quantum Approximate Optimization Algorithm** can also be used. Being locally simulated we are restricted by our machine capacity, so we need to decrease the size of the problem. The `pool` argument keeps only the most relevant features before building the QUBO.

```python
# Let's set the same starting point
myproject.preprocess()
features = myproject.feature_selection(max_cols=3, method="qaoa", pool=12)
model = myproject.evaluate("qsvc")
model.print_report()
```

Relevance, the mutual information between each feature and the target, is what `pool` uses to rank the features. It can be inspected directly.

```python
from falcondale.feature_selection import relevance

myproject.preprocess()
relevance(myproject.dataset).head(12)
```

With 12 candidates and 3 features to select there are only 220 options, so the **exact** solver can tell us the optimal selection for this QUBO. It is a good reference to check how close the other solvers got.

```python
myproject.preprocess()
myproject.feature_selection(max_cols=3, method="exact", pool=12)
```

By default the QUBO follows the mRMR formulation, where pairs of redundant features are penalized. The previous CMI formulation, rewarding complementary pairs, is still available with `formulation="cmi"`.

```python
myproject.preprocess()
myproject.feature_selection(max_cols=3, method="exact", pool=12, formulation="cmi")
```
